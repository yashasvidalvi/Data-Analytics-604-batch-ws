```python
import pandas as pd
```

```python
emp_data = {
    'emp_id':[101,102,103,104],
    'Gender':["Male","Female","Female","Male"],
    "Department":["HR","Placement","HR","Placement"],
    "Rating":['Avg','Good','Bad','Very Good']
}
emp_df = pd.DataFrame(emp_data)
emp_df
```

```python
from sklearn.preprocessing import LabelEncoder
label = LabelEncoder()
emp_df["Gender"]=label.fit_transform(emp_df[["Gender"]])
emp_df
```

```python
from sklearn.preprocessing import OrdinalEncoder
ordinal = OrdinalEncoder(categories=[['Avg','Good','Bad','Very Good']])
ordinal.fit_transform(emp_df[["Rating"]])
```
