```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
student_data = {
    'study_hours':[1,2,3,4,5,6],
    'marks':[30,40,50,60,70,80]
}
df= pd.DataFrame(student_data)
df
```

```python
x= df[['study_hours']]
y=df['marks']
y
```

```python
plt.plot(x,y)
```

```python
from sklearn.linear_model import LinearRegression
lmodel = LinearRegression()
lmodel.fit(x,y)
```

```python
lmodel.predict([[7]])
```

```python
lmodel.coef_
lmodel.intercept_
```
