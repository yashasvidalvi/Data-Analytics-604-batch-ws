```python
import pandas as pd
```

```python
error = {
    'y_act': [10,40,60,90],
    'y_pre' :[12,38,61,87]
}
df = pd.DataFrame(error)
df
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error,r2_score
print(mean_absolute_error(df['y_act'],df['y_pre']))
```

```python
mean_squared_error(df['y_act'],df['y_pre'])
```

```python
root_mean_squared_error(df['y_act'],df['y_pre'])
```

```python
r2_score(df['y_act'],df['y_pre'])
```
