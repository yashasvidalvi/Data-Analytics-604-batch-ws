```python
import pandas as pd
```

```python
student_df = pd.read_excel("student_data.xlsx")
student_df
```

```python
student_df['remaining_fees'] = student_df['fees_total']-student_df['fees_paid']
```

```python
student_df
```

```python
def student_performance(marks):
    if marks>=80:
        return "Excellent"
    elif marks>=60:
        return "Good"
    else:
        return "Poor"
per = student_df['marks'].apply(student_performance)
student_df['performance'] = per
student_df
```

```python
student_df['year'] = student_df['joining_date'].dt.year
```

```python
student_df
```

```python
from sklearn.preprocessing import LabelEncoder
label_en = LabelEncoder()
student_df['gender'] = label_en.fit_transform(student_df['gender'])
```

```python
student_df['gender'] = label_en.fit_transform(student_df['gender'])
student_df = pd.get_dummies(student_df,columns = ['course'])
student_df
```

```python
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()
student_df['age'] = std_scaler.fit_transform(student_df[['age']])
student_df
```

```python
from sklearn.preprocessing import StandardScaler
std_scaler = StandardScaler()
student_df[['study_hours',"attendance",'marks']] = std_scaler.fit_transform(student_df[['study_hours',"attendance",'marks']])
```
