```python
import pandas as pd
```

```python
student_df = pd.read_excel("student_data.xlsx")
student_df
```

```python
bins = [0,25,60,100]
lb = ["young","adult","Senior"]
student_df['age_gp']=pd.cut(student_df['age'],bins=bins,labels=lb)
```

```python
from sklearn.preprocessing import OrdinalEncoder
or_encoder = OrdinalEncoder(categories=[['young','adult','Senior']])
student_df['age_gp'] = or_encoder.fit_transform(student_df[['age_gp']])
```
