```python
import pandas as pd

dataset = {
    'gender': ["male", 'female', 'female', 'male', 'female'],
    'city': ['pune', 'mumbai', 'nashik', 'nagpur', 'nagar'],
    'p': ['yes', 'no', 'no', 'yes', 'no']
}

df = pd.DataFrame(dataset)
df
```

```python
from sklearn.preprocessing import LabelEncoder
label = LabelEncoder()
df['gender' ] = label.fit_transform(df['gender'])
df['city' ] = label.fit_transform(df['city'])
df['p' ] = label.fit_transform(df['p'])
df
```

```python
from sklearn.feature_selection import chi2
chi2(df[['gender','city']],df['p'])
```

```python
from sklearn.feature_selection import chi2
cs,pvalue = chi2(df[['gender','city']],df['p'])
data = pd.DataFrame({
    'feature':['gender','city'],
    'chi2 score':cs,
    'p_value':pvalue
})
data
```
