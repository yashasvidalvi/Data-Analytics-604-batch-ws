```python
import pandas as pd
from sklearn.cluster import DBSCAN

data = {
    "Point":["A","B","C","D","E","F","G","H","I","J"],
    "X":[1,1,2,2,3,8,8,9,9,20],
    "Y":[1,2,1,2,2,8,9,8,9,20]
}
df = pd.DataFrame(data)
df
```

```python
x = df[["X","Y"]]
model = DBSCAN(
    eps = 1.5,min_samples=4
)
df["Cluster"] = model.fit_predict(x)
df
```
