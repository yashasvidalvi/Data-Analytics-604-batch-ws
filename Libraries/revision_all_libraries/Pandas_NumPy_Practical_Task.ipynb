```python
# Sales Calculation
df['Gross_Amount'] = df['Quantity']*df["Unit_Price"]
```

```python
df[['Quantity','Unit_Price','Gross_Amount']].head()
```

```python
df['Discount_Amount'] = (df['Gross_Amount']*df['Discount']/100)
```

```python
df['Net_Sales'] = (df['Gross_Amount']-df['Discount_Amount'])
```

```python
df[['Gross_Amount','Discount_Amount','Net_Sales']].head()
```

```python
# Business Analysis
top_5_orders = df.sort_values(by='Net_Sales',ascending=False).head(5)
top_5_orders
```

```python
# NumPy Analysis
sales_array = np.array(df['Net_Sales'])
sales_array
```

```python
df['Sales_Status'] = np.where(df['Net_Sales']>=average_sales,'High Value','Regular')
```
