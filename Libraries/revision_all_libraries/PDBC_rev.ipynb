```python
#Task 6
total_salary = {
    "IT": 0,
    "HR": 0,
    "Sales": 0
}

count = {
    "IT": 0,
    "HR": 0,
    "Sales": 0
}
```

```python
cursor.execute("SELECT * FROM employees")

records = cursor.fetchall()
for row in records:

    department = row[2]
    salary = float(row[3])

    total_salary[department] += salary
    count[department] += 1
for department in total_salary:

    average = total_salary[department] / count[department]

    print(
        department,
        "Total:", total_salary[department],
        "Average:", average
    )
```

```python
#Task 8

cursor.execute("SELECT * FROM employees")

records = cursor.fetchall()

for row in records:

    emp_id = row[0]
    department = row[2]
    salary = float(row[3])

    if department == "IT":
        increment_rate = 12

    elif department == "Sales":
        increment_rate = 10

    else:
        increment_rate = 8

    new_salary = salary + (salary * increment_rate / 100)
```

```python
query = """
    UPDATE employees
    SET salary = %s
    WHERE emp_id = %s
    """

values = (new_salary, emp_id)

cursor.execute(query, values)
con.commit()
print("Salary Updated Successfully")
```

```python
#Task 10
cursor.execute("SELECT * FROM employees")

records = cursor.fetchall()

highest = records[0]
lowest = records[0]
for row in records:

    if row[3] > highest[3]:
        highest = row

    if row[3] < lowest[3]:
        lowest = row
difference = highest[3] - lowest[3]
print("Highest Salary Employee:", highest[1])
print("Highest Salary:", highest[3])

print("Lowest Salary Employee:", lowest[1])
print("Lowest Salary:", lowest[3])

print("Difference:", difference)
```
