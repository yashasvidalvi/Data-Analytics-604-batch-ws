```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("data_cleaning_20_records.xlsx")
df
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['Customer_Name'] = df['Customer_Name'].str.strip()
df['City'] = df['City'].str.strip()
df['Course'] = df['Course'].str.strip()
df['Payment_Status'] = df['Payment_Status'].str.strip()
```

```python
df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['Course'] = df['Course'].str.title()
df['Payment_Status'] = df['Payment_Status'].str.title()
```

```python
df['Customer_Name'] = (df['Customer_Name'].str.replace(r"[^A-Za-z ]","",regex=True).str.replace(r"\s+"," ",regex=True).str.strip())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df.loc[(df["Age"]<18) | (df["Age"]>60),"Age"]=np.nan
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Fees"] = df["Fees"].fillna(df["Fees"].median())
```

```python
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"],errors="coerce",dayfirst=False)
df["Admission_Date"]
```

```python
df[df["Admission_Date"].isna()]
```

```python
df["Phone"] = (
    df["Phone"]
    .astype(str)
    .str.replace(r"[\s-]", "", regex=True)
)
df
```

```python
df["Phone_Valid"] = df["Phone"].str.match(r"^[6-9]\d{9}$")
print(df[["Phone","Phone_Valid"]])
```

```python
df["Payment_Status"] = (df["Payment_Status"].str.strip().str.title())
df["Payment_Status"].value_counts()
```

```python
df.duplicated().sum()
```

```python
df[df["Customer_ID"].duplicated(keep=False)]
```

```python
df["Age_group"] = pd.cut(
    df["Age"],bins=[0,25,35,45,100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46+"
    ]
)
df
```

```python
df["Fee_Category"] = pd.cut(
    df["Fees"],
    bins=[0,25000,40000,60000],
    labels=[
        "Low",
        "Medium",
        "High"
    ]
)
df
```

```python
df = df[df["Phone_Valid"]==True]
```

```python
df.drop(columns=["Phone_Valid"], inplace=True)
```

```python
df.reset_index(drop=True,inplace=True)
```
