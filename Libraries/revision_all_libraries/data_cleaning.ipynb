```python
df['Customer_Name'] = df['Customer_Name'].str.strip()
df['City'] = df['City'].str.strip()
df['Course'] = df['Course'].str.strip()
df['Payment_Status'] = df['Payment_Status'].str.strip()
```

```python
df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['Course'] = df['Course'].str.title()
df['Payment_Status'] = df['Payment_Status'].str.title()
```

```python
df['Customer_Name'] = (df['Customer_Name'].str.replace(r"[^A-Za-z ]","",regex=True).str.replace(r"\s+"," ",regex=True).str.strip())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df.loc[(df["Age"]<18) | (df["Age"]>60),"Age"]=np.nan
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Fees"] = df["Fees"].fillna(df["Fees"].median())
```

```python
df["Admission_Date"] = pd.to_datetime(df["Admission_Date"],errors="coerce",dayfirst=False)
df["Admission_Date"]
```
