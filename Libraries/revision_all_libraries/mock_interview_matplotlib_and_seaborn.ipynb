```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "Department":["IT","HR","IT","HR","Sales","Sales","IT","HR"],
    "Salary": [50000,30000,60000,40000,35000,45000,70000,50000],
})
df
```

```python
sns.barplot(data=df,x="Department",y="Salary",estimator="mean")
plt.title("Average Salary By Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.show()
```

```python

import matplotlib.pyplot as plt

departments = ['HR', 'IT', 'Sales', 'Finance', 'Marketing']
employees = [10, 25, 20, 15, 18]

plt.bar(departments, employees)
plt.title('Number of Employees in Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.show()
```
