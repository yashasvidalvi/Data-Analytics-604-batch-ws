```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'],errors='coerce')
```

```python
plt.Figure(figsize=(8,5))
sns.countplot(data=df,x='Product_Category')
plt.title("Number of orders by product category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(8,5))
sns.countplot(data=df,x='Customer_Segment')
plt.title("Number of orders by Customer_Segment")
plt.xlabel("Customer_Segment")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(10,5))
sns.countplot(data=df,x='Product_Category',hue='Customer_Segment')
plt.title("Orders by Product_Category and Customer_Segment")
plt.xlabel("Product_Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.legend(title="Customer Segment")
plt.tight_layout()
plt.show()
```

```python
sns.barplot(data=df,x="Product_Category",y="Sales_Amount")
plt.title("Average Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sns.barplot(data=df,x="Region",y="Profit")
plt.xlabel("Region")
plt.ylabel("Average Profit")
plt.title("Average Profit by Region")
plt.show()
```

```python
sns.barplot(data=df,x="Product_Category",y="Profit",hue="Customer_Segment")
plt.title("Average Profit by Product_Category")
plt.xlabel("Product_Category")
plt.ylabel("Profit")
plt.legend(title="Customer Segment")
plt.show()
```

```python
sns.histplot(data=df,x="Sales_Amount",bins=30)
plt.title("Distribution of Sales Amount")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(data=df,x="Profit")
plt.title("Distribution of Profit")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(data=df,x="Sales_Amount",kde=True)
plt.title("Sales Distribution with KDE")
plt.xlabel("Sales Amount")
plt.ylabel("Frquency")
plt.show()
```

```python
sns.boxplot(data=df,y="Sales_Amount")
plt.title("Sales Amount Outlier Analysis")
plt.ylabel("Sales Amount")
plt.show()
```

```python
sns.boxplot(data=df,x="Product_Category",y="Sales_Amount")
plt.title("Sales Variation by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount")
plt.show()
```

```python
sns.boxplot(data=df,x="Customer_Segment",y="Profit")
plt.title("Profit Distribution by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Profit")
plt.show()
```

```python
sns.boxplot(data=df,x="Sales_Channel",y="Delivery_Days")
plt.title("Delivery Performance by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Delivery Days")
plt.show()
```

```python
sns.scatterplot(data=df,x="Marketing_Spend",y="Sales_Amount")
plt.title("Marketing Spend vs Sales Amount")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales Amount")
plt.show()
```

```python
sns.scatterplot(data=df, x="Units_Sold",y="Sales_Amount")
plt.title("Units Sold vs Sales Amount")
plt.xlabel("Units Sold")
plt.ylabel("Sales Amount")
plt.show()
```

```python
sns.scatterplot(data=df,x="Marketing_Spend",y="Sales_Amount",hue="Product_Category")
plt.title("Marketing Spend vs Sales by Category")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales Amount")
plt.legend(title="Product Category")
plt.show()
```

```python
sns.scatterplot(data=df,x="Marketing_Spend",y="Sales_Amount",hue="Product_Category",size="Units_Sold")
plt.title("Marketing, Sales, Category and Units Sold")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales Amount")
plt.show()
```

```python
daily_Sales = (df.groupby("Order_Date",as_index=False)["Sales_Amount"].sum())
sns.lineplot(data=daily_Sales,x="Order_Date",y="Sales_Amount")
plt.title("Daily Sales Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
category_sales = (df.groupby(['Order_Date','Product_Category'],as_index=False)["Sales_Amount"].sum())
sns.lineplot(data=category_sales,x="Order_Date",y="Sales_Amount",hue="Product_Category")
plt.title("Daily Sales Trend by Product Category")
plt.xlabel("Order Date")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
daily_profit = (df.groupby("Order_Date",as_index=False)["Profit"].sum())
sns.lineplot(data=daily_profit,x="Order_Date",y="Profit")
plt.title("Daily Profit Trend")
plt.xlabel("Order Date")
plt.ylabel("Total Profit")
plt.show()
```
