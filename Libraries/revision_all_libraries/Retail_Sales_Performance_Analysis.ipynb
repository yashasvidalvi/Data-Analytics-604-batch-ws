```python
conditions = [df["Sales"]<10000,(df["Sales"]>=10000)&(df["Sales"]<=30000),df["Sales"]>30000]
choices = ["Low","Medium","High"]
df["Sales_Category"] = np.select(conditions,choices,default="Unknown")
df[["Sales","Sales_Category"]].head(20)
```

```python
con =  [df["Profit"]<2000,(df["Profit"]>=2000)&(df["Profit"]<=5000),df["Profit"]>5000]
choice = ["Low Profit","Medium Profit","High Profit"]
df["Profit_Category"] = np.select(con,choice,default = "Unknown")
df[["Profit","Profit_Category"]].head(20)
```

```python
plt.plot(daily_sales.index,daily_Profit.values)
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_sales = (df.groupby("Product")["Sales"].sum().sort_values(ascending=False))
bars = plt.bar(product_sales.index,product_sales.values)
plt.title("Product Vs Total Sales")
plt.xlabel("product")
plt.ylabel("Total Sales")
plt.bar_label(bars)
plt.show()
```

```python
Region_Sales = (df.groupby("Region")["Sales"].sum().sort_values(ascending=False))
best_region = Region_Sales.idxmax()
best_sales = Region_Sales.max()
bars = plt.bar(Region_Sales.index,Region_Sales.values)
plt.title("Region Vs Total Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.bar_label(bars)
plt.show()
print(best_region, best_sales)
```

```python
Profit_sales = (df.groupby("Product")["Profit"].sum().sort_values(ascending=False))
bars = plt.bar(Profit_sales.index,Profit_sales.values)
plt.title("Product Vs Total Profit")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.bar_label(bars)
plt.show()
```
