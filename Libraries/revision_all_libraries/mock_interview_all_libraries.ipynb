```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("Monthly_Sales_and_Student_Marks.xlsx")
df
```

```python
sns.lineplot(data=df,x="Month",y="Sales")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
df1 = pd.read_excel("Monthly_Sales_and_Student_Marks.xlsx",sheet_name="Student_Marks")
sns.histplot(data=df1,x="Marks",bins=5)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```
