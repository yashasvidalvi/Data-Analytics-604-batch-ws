```python
data = {
    "Employee": ["Amit", "Priya", "Rahul", "Sneha", "Raj"],
    "Department": ["IT", "HR", "IT", "Testing", "Testing"],
    "Salary": [45000, 38000, 60000, 42000, 50000],
    "Experience": [2, 3, 5, 2, 4]
}
df = pd.DataFrame(data)
df
```

```python
avg = df["Salary"].mean()
avg
```

```python
df[df["Salary"]>avg]
```

```python
arr3 = np.array(df["Salary"])
print(arr3.max())
print(arr3.min())
print(arr3.mean())
```

```python
max = df["Salary"].max()
df[df["Salary"]==max]
```

```python
df["Department"].value_counts()
```

```python
sns.countplot(x="Department",data=df)
```

```python
plt.bar(df["Employee"],df["Salary"],data=df)
```

```python
df["Salary_category"] = np.where(
    df["Salary"]>=50000,
    "High",
    np.where(
        df["Salary"]>=40000,
        "Medium",
        "Low"
    )
)
df
```

```python
filter_df = df[(df["Salary"]>40000)&(df["Experience"]>=3)]
filter_df
```

```python
sns.barplot(x="Employee",y="Salary",data=filter_df)
```
