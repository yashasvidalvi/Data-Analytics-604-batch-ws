```python
import numpy as np
```

```python
arr1 = np.array([10,20,30,40,50])
print(arr1)
```

```python
#indexing
print(arr1[1])
print(arr1[-2])
```

```python
arr2 = np.array([[11,22,33],[44,55,66]])
print(arr2)
```

```python
print(arr2[0])
print(arr2[1][1])
```

```python
print(arr2[0,1])
```

```python
arr2 = np.array([[11,22,33],[44,55,66],[77,88,99],[1,2,3]])
print(arr2)
```

```python
#slicing
arr1 = np.array([11,22,33,44,55])
print(arr1)
print(arr1[-2])
print(arr1[1:-1:1])
print(arr1[-1:-4:-1])
```
