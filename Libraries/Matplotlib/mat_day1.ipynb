```python
# from matplotlib import pyplot as plt
import matplotlib.pyplot as plt
```

```python
month = ["jan","feb","mar","apr"]
sales = [10000,20000,50000,70000]
plt.plot(month,sales)
plt.show()
```

```python
month = ["jan","feb","mar","apr"]
sales = [10000,20000,50000,70000]
plt.plot(month,sales,c = "red",lw=5,ls="dashed")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("MonthWise Sales")
plt.yticks([10000,15000,20000,25000,30000,35000,40000,45000,50000,55000,60000,65000,70000])
plt.grid()
plt.show()
```

```python
month = ['jan','feb','mar','apr']
no_of_admission = [12,15,34,23]
plt.plot(month,no_of_admission,c = "pink",lw = 10,ls = "--")
plt.grid(visible=True,which='major',axis="y")
plt.show()
```

```python
month = ["jan","feb","mar","apr"]
sales = [10000,20000,50000,70000]
plt.plot(month,sales,marker = "o",mec = "red",mfc = 'green')
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("MonthWise Sales")
plt.grid()
plt.show()
```

```python
month = ["jan","feb","mar","apr"]
sales = [10000,20000,50000,70000]
plt.plot(month,sales,marker = "o",mec = "red",mfc = 'green')
plt.xlabel("Month",loc = "center",color = "red",fontsize = 20)
plt.ylabel("Sales",loc = "center",color = "green",fontsize = 20)
plt.title("MonthWise Sales",loc = "center",color = "yellow",fontsize = 20)
plt.show()
```

```python
import pandas as pd
sales_df = pd.read_csv("sales_data.csv")
sales_df
```

```python
month = sales_df["month"]
sales_2024 = sales_df["sales_2024"]
sales_2025 = sales_df["sales_2025"]
```

```python
plt.plot(month,sales_2024, label = "2024 sales")
plt.plot(month,sales_2025,label = "2025 sales")
plt.legend()
```

```python
courses = ["Python Development","Data Analytics","Java development","Testing"]
no_students = [34,85,53,56]
```

```python
plt.bar(courses,no_students,color = "green")
plt.xlabel("courses")
plt.ylabel("No of students")
plt.show()
```

```python
plt.barh(courses,no_students,color = "green")
plt.xlabel("courses")
plt.ylabel("No of students")
plt.show()
```
