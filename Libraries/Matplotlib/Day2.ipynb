```python
import matplotlib.pyplot as plt
```

```python
#histogram
marks = [10,15,17,25,35,40,55,69,75,80,91,95,99,93]
bins = [0,30,60,100]
plt.hist(marks,bins=bins)
plt.xticks(bins)
plt.show()
```

```python
#scatter plot
study_hours = [1,2,3,4,5,6]
marks = [40,55,60,70,80,99]
plt.scatter(study_hours,marks)
plt.show()
```

```python
#subplot
month = ['jan','feb','mar','apr']
sales_2023 = [10,15,17,20]
sales_2024 = [15,25,37,40]
plt.subplot(1,2,1)
plt.plot(month,sales_2023)
plt.subplot(1,2,2)
plt.plot(month,sales_2024)
plt.show()
```

```python
month = ['jan','feb','mar','apr']
sales_2023 = [10,15,17,20]
sales_2024 = [15,25,37,40]
plt.subplot(2,1,1)
plt.plot(month,sales_2023)
plt.title("sales_2023")
plt.xlabel("Month")
plt.ylabel("sales")
plt.subplot(2,1,2)
plt.plot(month,sales_2024)
plt.title("sales_2024")
plt.show()
```
