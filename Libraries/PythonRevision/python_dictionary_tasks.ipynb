```python
products = {
 "P101": {"name": "Laptop", "price": 55000, "stock": 8},
 "P102": {"name": "Mouse", "price": 700, "stock": 25},
 "P103": {"name": "Keyboard", "price": 1200, "stock": 15},
 "P104": {"name": "Monitor", "price": 14000, "stock": 5},
 "P105": {"name": "Headphones", "price": 2500, "stock": 0}
}
products
```

```python
for pro in products.values():
    print({pro["name"]},{pro["price"]})
```

```python
for p in products.values():
    if p["stock"]<10:
        print(p['name'])
```

```python
for p in products.values():
    if p["stock"]==0:
        print(p["name"])
```

```python
for pid,p in products.items():
    total_val = p["price"]*p["stock"]
    print(f"{p['name']} ({pid}) Total Value: {total_val}")
```

```python
most_expensive = max(products.values(),key=lambda p:p["price"])
most_expensive["name"]
```

```python
cheapest = min(products.values(),key=lambda p:p["price"])
cheapest["name"]
```

```python
available_products = [p["name"] for p in products.values() if p["stock"]>0]
available_products 
```

```python
new_dic = {p["name"]:p["stock"] for p in products.values()}
new_dic
```

```python
products["P104"]["stock"]=12
products
```

```python
for p in products.values():
    if p["price"]>10000:
        p["price"] = p["price"]*0.95
print(products)
```
