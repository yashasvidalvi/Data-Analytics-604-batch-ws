```python
students = ["Rahul","Priya","Amit","Sneha","Raj","Anjali"]
print(students)
```

```python
for i in students:
    if i.startswith("A"):
        print(i)
```

```python
for i in students:
    if len(i)>5:
        print(i)
```

```python
for i in students:
    print(i.upper())
```

```python
students.sort()
print(students)
```

```python
long = max(students,key=len)
print(long)
```

```python
if "Sneha" in students:
    print("Sneha exists")
else:
    print("Sneha not exists")
```

```python
new_list = {name: len(name) for name in students}
print(new_list)
```

```python
students.remove("Raj")
print(students)
```
