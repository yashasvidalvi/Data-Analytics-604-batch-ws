```python
import pandas as pd
import seaborn as sns
```

```python
student_df  = pd.read_csv("admission_data.csv")
student_df
```

```python
import matplotlib.pyplot as plt
sns.barplot(student_df,x = "branch_name",y = "paid_fees",estimator=sum,hue = 'course',width = 0.5,gap = 0.2)
plt.title("Branch-wise collection")
plt.grid()
```

```python
df = pd.read_csv("sales.csv")
gp = df.groupby("month")
result = gp["sales_amount"].sum()
```

```python
month = result.index
sales = result.values
```

```python
import matplotlib.pyplot as plt
plt.plot(month,sales)
```

```python
sns.lineplot(data=df,x="month",y="sales_amount",estimator="sum",hue="course")
```
