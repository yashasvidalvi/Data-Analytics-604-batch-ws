```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("admission_data.csv")
bname = df['branch_name']
bname
paid_fees = df['paid_fees']
paid_fees
```

```python
plt.bar(bname,paid_fees)
plt.show()
```

```python
gp = df.groupby(by="branch_name")
result = gp["paid_fees"].sum()
courses = result.index
sales_amount = result.values
```

```python
plt.bar(courses,sales_amount)
```

```python
gp = df.groupby(by="course")
result = gp["paid_fees"].sum()
courses = result.index
sales_amount = result.values
```

```python
plt.bar(courses,sales_amount)
```
