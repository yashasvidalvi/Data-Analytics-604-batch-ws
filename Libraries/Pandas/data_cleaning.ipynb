```python
import pandas as pd
```

```python
student_df = pd.read_excel("student_records_20.xlsx")
student_df
```

```python
#null values are check
student_df.isnull().sum()
```

```python
#dropna()
df = student_df["name"].dropna()
# df.isnull().sum()
df
```

```python
student_df["name"].fillna("Unknown")
```

```python
course = student_df["course"].mode()
student_df["course"].fillna(course)
student_df
```

```python
branch = student_df["branch"].mode()
student_df["branch"].fillna(branch)
```

```python
student_df['course'].fillna(method = 'ffill')
```

```python
avg = student_df['fees'].mean()
student_df['fees'].fillna(avg,inplace=True)
```
