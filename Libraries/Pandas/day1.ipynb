```python
employee = {
    "empid": [101,102,103,104,105],
    "empname": ["Siya","Shreya","Diksha","Aarya","Avni"],
    "department":["HR","IT","Finance","IT","Marketing"],
    "basic_salary":[50000,40000,70000,80000,60000],
}
df = pd.DataFrame(employee)
df
```

```python
new = pd.DataFrame(
    [
        {
            "empid":106,
            "empname":"Piya",
            "department":"Operation",
            "basic_salary":50000
        }
    ]
)
df = pd.concat([df,new],ignore_index=True)
df
```

```python
df["HRA_10_percent"] = df["basic_salary"] * 0.10
df["DA_8_percent"] = df["basic_salary"] * 0.08
df
```

```python
df["salary"] = df["basic_salary"]+df["HRA_10_percent"]+df["DA_8_percent"]
df
```

```python
df = df[df["empid"] != 104].reset_index(drop=True)
df
```

```python
df = df.drop(columns=["department"])
df
```
